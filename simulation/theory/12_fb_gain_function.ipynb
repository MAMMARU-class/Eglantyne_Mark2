```python
import numpy as np
import matplotlib.pyplot as plt

# 係数の範囲を定義 (-100から100まで)
acc_fb_coeff = np.linspace(-50, 50, 400)

# C++の式を再現
# acc_fb = 2.0 / PI * atan(acc_fb_coeff) + 1.0
acc_fb = (2.0 / np.pi) * np.arctan(acc_fb_coeff) + 1.0

# グラフ作成
plt.figure(figsize=(10, 6))
plt.plot(acc_fb_coeff, acc_fb, label=r'$y = \frac{2}{\pi} \arctan(x) + 1$', color='blue', linewidth=2)

# 補助線の追加
plt.axhline(y=1.0, color='gray', linestyle='--', alpha=0.5, label='Center (1.0)')
plt.axhline(y=0.0, color='red', linestyle=':', label='Lower Bound (0.0)')
plt.axhline(y=2.0, color='green', linestyle=':', label='Upper Bound (2.0)')
plt.axvline(x=0.0, color='black', linewidth=0.8)

# ラベルと設定
plt.title('Acceleration Feedback Gain Curve (C++ Implementation)', fontsize=14)
plt.xlabel('acc_fb_coeff', fontsize=12)
plt.ylabel('acc_fb (Output Gain)', fontsize=12)
plt.grid(True, which='both', linestyle='--', alpha=0.5)
plt.legend()

plt.show()
```
